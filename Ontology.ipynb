```python
from owlready2 import ObjectProperty, Thing, get_ontology, sync_reasoner
```

```python
# ==================================================
# 1. 生物オントロジー (ネームスペース A)
# ==================================================
bio_onto = get_ontology("http://example.org/biology.owl")

with bio_onto:
    class Snake(Thing):  # ニシキヘビ（生物）クラス
        pass
    class Food(Thing) : # 餌 クラス
        pass
    class eats(ObjectProperty):  # myPython.eats = [ ]
        domain = [Snake]
        range = [Food]
```

```python
    myPython = Snake("python")
    hisFood = Food("chickenEggs")
    print(myPython)            # biology.python (myPython という変数に入っているデータは、biology というオントロジーに存在する "python" という個体)
    print(type(myPython))      # biology.Snake (myPython という変数に入っているデータは、biology というオントロジーに存在する "Snake" というクラス)
    print(hisFood)             # biology.chickenEggs
    print(type(hisFood))       # biology.Food
    print(myPython.eats)       # []
    print(type(myPython.eats)) # <class 'owlready2.prop.IndividualValueList'>
    myPython.eats.append(hisFood)
    myPython.eats.append(Food("Frogs"))
    print(myPython.eats)       # [biology.chickenEggs, biology.Frogs]
```

```python
# --------------------------------------------------
# 個体を作成（あえてクラスを指定せず、ただの Thing として作る）
# --------------------------------------------------

with bio_onto:
    # somethingA も somethingB も、この時点では「Snake」や「Food」ではなく単なる「Thing」
    somethingA = Thing("ABC")
    somethingB = Thing("XYZ")

    # 関係だけを接続する
    somethingA.eats.append(somethingB)
```

```python
# --- 推論前の確認 ---
print("=== 推論前 ===")
print(f"somethingA の所属クラス: {somethingA.is_a}")  # [owl.Thing]
print(f"somethingB の所属クラス: {somethingB.is_a}")  # [owl.Thing]
```

```python
# --------------------------------------------------
# 推論エンジン（sync_reasoner）の実行
# --------------------------------------------------
sync_reasoner(bio_onto)
```

```python
print("\n=== 推論後 ===")
print(f"somethingA の所属クラス: {somethingA.is_a}")
# 出力: [owl.Thing, biology.Snake]  (eatsの主語だから Snake だと判定された！)

print(f"somethingB の所属クラス: {somethingB.is_a}")
# 出力: [owl.Thing, biology.Food]   (eatsの目的語だから Food だと判定された！)
```

```python
# ==================================================
# 2. IT/プログラミングオントロジー (ネームスペース B)
# ==================================================
it_onto = get_ontology("http://example.org/technology.owl")

with it_onto:
    class Language(Thing):  # プログラミング言語クラス
        pass
    class OperatingSystem(Thing):  # OSクラス
        pass
    class runsOn(ObjectProperty):  # myPython.runsOn = [ ]
        domain = [Language]
        range = [OperatingSystem]
```

```python
    myPython = Language("python3")
    myOS = OperatingSystem("Linux")
    myPython.runsOn.append(myOS)
```

```python
print(myPython)
print(myPython.runsOn)
```

```python
# --------------------------------------------------
# 個体を作成（あえてクラスを指定せず、ただの Thing として作る）
# --------------------------------------------------

with it_onto:
    # somethingC も somethingD も、この時点では「Language」や「OperatingSystem」ではなく単なる「Thing」
    somethingC = Thing("EFG")
    somethingD = Thing("UVW")

    # 関係だけを接続する
    somethingC.runsOn.append(somethingD)
```

```python
# --- 推論前の確認 ---
print("=== 推論前 ===")
print(f"somethingC の所属クラス: {somethingC.is_a}")  # [owl.Thing]
print(f"somethingD の所属クラス: {somethingD.is_a}")  # [owl.Thing]
```

```python
# --------------------------------------------------
# 推論エンジン（sync_reasoner）の実行
# --------------------------------------------------
sync_reasoner(it_onto)
```

```python
print("\n=== 推論後 ===")
print(f"somethingC の所属クラス: {somethingC.is_a}")
# 出力: [owl.Thing, technology.Language]  (runsOnの主語だから Language だと判定された！)

print(f"somethingD の所属クラス: {somethingD.is_a}")
# 出力: [owl.Thing, technology.OperatingSystem]   (runsOnの目的語だから OperatingSystem だと判定された！)
```
